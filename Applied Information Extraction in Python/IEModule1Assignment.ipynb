# Task 1: Create reg expression

These are possible formats a phone number could be stored in:

Parentheses: (409) 339-9584

Dashes: 943-641-7270

Dots: 574.874.7163

Spaces: 175 335 3625

International: +1-536-234-1765

Plain: 6565865784

Create a 'reg' expression such that the given code is able to extract phone numbers in every one of these formats.

When we use reg.findall, we want to obtain a list of 3 strings containing the digits of each phone number separated by section. For example, if we have:

`str = John Smith: 123 456 7890`

We want `re.findall(reg, str)` on this line using our regex expression to obtain a list that contains the following:

['123','456','7890']

Look up how to use regex 'capturing groups' to accomplish this.

Assign your reg expression to `reg` in `task_1_solution`.

We use your reg expression in `extract_contents` to extract the contents of `lines` into a dictionary where each entry is formatted like so:

`'First Last': ['000', '000', '0000']`

````python
import re


def extract_contents(reg):
    extracted_contents = {}
    for line in lines:
        line_content = line.split(':')
        matches = re.findall(reg, line_content[1])
        for match in matches:
            
            # Filter out empty strings and None values from the match
            parts = [part for part in match if part]
            extracted_contents[line_content[0]] = (parts)
    return extracted_contents
````

````python

def task_1_solution():

    # Regular expression to match all the required formats
    reg_str = r"""
        (?:\+1[-\s\.]?)?                # optional country code
        (?:\(?(\d{3})\)?[-\s\.]?)       # area code
        (\d{3})                         # central office code
        [-\s\.]?                        # separator
        (\d{4})                         # line number
    """
    reg = re.compile(reg_str, re.VERBOSE)
    return extract_contents(reg), reg
````

# Task 2: Normalizing Data

Now, we'd like to normalize the phone numbers to the format: (123) 456-7890.

The function below will take in the ```stu_contents``` that you helped define in task 1 as an argument. It will then alter every entry in the ```stu_contents``` dictionary to be a string in the (123) 456-7890 format. The function will return the modified ```extracted_contents``` dictionary.

For example, 

The first entry in your task 1 `extracted_contents` is 

'David Smith': ['838', '177', '8461']

This should be turned into

'David Smith': '(838) 177-8461'

````python
import copy
````

````python

def task_2_solution():

    extracted_contents, _ = task_1_solution()

    extracted_contents = copy.deepcopy(extracted_contents)
    
    # YOUR CODE HERE
    for name, parts in extracted_contents.items():
        if len(parts) == 3:
            area, mid, last = parts
            formatted = f"({area}) {mid}-{last}"
            extracted_contents[name] = formatted
        else:
            # If the phone number format was invalid and didn't capture 3 parts
            extracted_contents[name] = "Invalid Number"

    return extracted_contents
````
